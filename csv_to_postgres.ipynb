```python
import pandas as pd
from sqlalchemy import create_engine

# csv dosyasını pandas ile oku
csv_yolu = "WA_Fn-UseC_-HR-Employee-Attrition.csv"
df = pd.read_csv(csv_yolu)

# kolon isimlerini sql formatına uygun hale getirmek iyi bir pratiktir (boşlukları sil, küçük harf yap)
df.columns = [col.lower().replace(' ', '_') for col in df.columns]

# postgresql bağlantısını kurma
# format: postgresql://kullanici_adi:sifre@host:port/veritabani_adi
kullanici_adi = "postgres"
sifre = ""
host = "localhost"
port = "5432"
veritabani_adi = "ik_projesi_db" # bu veritabanını pgAdmin'de önceden açmış olmamız gerek.

engine = create_engine(f'postgresql+psycopg2://{kullanici_adi}:{sifre}@{host}:{port}/{veritabani_adi}')

# veriyi veritabanına yaz
tablo_adi = "employee_attrition"

# if_exists='replace' -> tablo varsa silip yenisini kurar (append dersek üzerine ekler)
df.to_sql(tablo_adi, engine, if_exists='replace', index=False)

print(f"{len(df)} satır veri PostgreSQL'deki '{tablo_adi}' tablosuna başarıyla yüklendi.")
```
