```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC
from sklearn.ensemble import GradientBoostingClassifier, RandomForestClassifier 

from sklearn.model_selection import train_test_split

from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
```

```python
y = df_hazir["isten_ayrilma"]
x = df_hazir.drop("isten_ayrilma" , axis =1)
```

```python

X_train , X_test , y_train , y_test = train_test_split( x,
                                                        y,
                                                        test_size = 0.2,
                                                        random_state = 42,
                                                        stratify = y    
                                                    )
```

```python
print(f"Eğitim seti satır sayısı (x_train) = {X_train.shape[0]}" " ,Oran : " , 1176/1470)
print(f"Test seti satır sayısı (x_test) = {X_test.shape[0]}" " ,Oran : " , 294/1470)
print(f"Toplam özellik (Sütun) sayısı = {X_train.shape[1]}")


print(y_test.shape)

print(y_train.shape) 
```

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train) # hem öğren , hem dönüştür
X_test_scaled = scaler.transform(X_test) # sadece dönüştür
```

```python
modeller = {
    "Logistic Regression": LogisticRegression(max_iter=1000, random_state=42),
    "Decision Tree": DecisionTreeClassifier(random_state=42),
    "Random Forest": RandomForestClassifier(n_estimators=100, random_state=42),
    "SVM (Destek Vektör)": SVC(random_state=42),
    "Gradient Boosting": GradientBoostingClassifier(random_state=42)
}
```

```python
sonuclar = []

for isim , model  in modeller.items():
    model.fit(X_train_scaled , y_train)
    y_pred = model.predict(X_test_scaled)

    sonuclar.append({
        "Model" : isim,
        "Accuracy" : accuracy_score(y_test , y_pred),
        "Precision" : precision_score(y_test , y_pred , zero_division = 0),
        "Recall" : recall_score(y_test , y_pred , zero_division =0),
        "F1-Score" : f1_score(y_test , y_pred , zero_division = 0)
    })
```

```python
karsilastirma_df = pd.DataFrame(sonuclar).sort_values(by="F1-Score", ascending=False).round(3)
print(karsilastirma_df)
```

```python
import warnings
warnings.filterwarnings('ignore', category=FutureWarning)


from sklearn.model_selection import GridSearchCV
from sklearn.metrics import roc_auc_score

# pandas'ın tabloyu yazdırırken uzun parametre metinlerini "..." diyerek kesmemesi için bir ayar:
pd.set_option('display.max_colwidth', None)

# her model için özel hiperparametreler
# sözlük yapısı: "Model İsmi": (modelin Kendisi, denenecek ayarlar)
modeller_ve_parametreler = {
    "Logistic Regression": (
        LogisticRegression(max_iter=1000, random_state=42),
        {'C': [0.01, 0.1, 1, 10], 'class_weight': ['balanced']}
    ),
    "Decision Tree": (
        DecisionTreeClassifier(random_state=42),
        {'max_depth': [5, 10, 15], 'min_samples_split': [2, 5, 10], 'class_weight': ['balanced']}
    ),
    "Random Forest": (
        RandomForestClassifier(random_state=42),
        {'n_estimators': [100, 200], 'max_depth': [5, 10, 15], 'class_weight': ['balanced']}
    ),
    "SVM (Destek Vektör)": (
        # SVM'in roc-auc için olasılık (predict_proba) üretebilmesi için probability=True ŞARTTIR!
        SVC(random_state=42, probability=True), 
        {'C': [0.1, 1, 10], 'kernel': ['linear', 'rbf'], 'class_weight': ['balanced']}
    ),
    "Gradient Boosting": (
        GradientBoostingClassifier(random_state=42),
        # Gradient Boosting kütüphanesinde 'class_weight' parametresi yok. 
        # Onun yerine öğrenme hızını (learning_rate) ve ağaç sayısını optimize ediyoruz.
        {"n_estimators": [100, 200], "max_depth": [3, 5], "learning_rate": [0.01, 0.1]} 
    )
}

optimizasyon_sonuclari = []


for isim, (model, parametreler) in modeller_ve_parametreler.items():
    
    # her model için ayrı GridSearch kur
    grid = GridSearchCV(
        estimator=model,
        param_grid=parametreler,
        cv=5,
        scoring='roc_auc',
        n_jobs=-1 # Tüm işlemci gücünü kullan
    )
    
    
    grid.fit(X_train_scaled, y_train)
    
    # O modelin en iyi halini (best estimator) al
    en_iyi_model = grid.best_estimator_
    
    # olasılıkları tahmin et ve roc-auc hesapla
    y_tahmin_olasilik = en_iyi_model.predict_proba(X_test_scaled)[:, 1]
    roc_skoru = roc_auc_score(y_test, y_tahmin_olasilik)
    
    
    optimizasyon_sonuclari.append({
        "Model": isim,
        "roc-auc skoru": roc_skoru,
        "En İyi Ayarlar": grid.best_params_
    })
    


rovanş_df = pd.DataFrame(optimizasyon_sonuclari).sort_values(by="roc-auc skoru", ascending=False).round(3)


print("="*100)
print(rovanş_df.to_string(index=False))
```

```python
from sklearn.metrics import confusion_matrix, classification_report

# en optimize ayarlar ile modelleri inceliyoruz
optimize_modeller = {
    "Gradient Boosting": GradientBoostingClassifier(learning_rate = 0.1, max_depth = 3, n_estimators = 200, random_state = 42),
    "Logistic Regression": LogisticRegression(C = 10, class_weight = "balanced", max_iter = 1000, random_state = 42),
    "Random Forest": RandomForestClassifier(class_weight = "balanced", max_depth = 15, n_estimators = 100, random_state = 42),
    "SVM (Destek Vektör)": SVC(C = 1, class_weight = "balanced", kernel = "rbf", random_state = 42)
}

# her model için confusıon matrıx oluşturuyoruz
for isim, model in optimize_modeller.items():
    
    model.fit(X_train_scaled, y_train)
    
    y_pred = model.predict(X_test_scaled)
    
    print(f"{isim} modeli")
    print(f"{'='*50}")
    
    cm = confusion_matrix(y_test, y_pred)
    print("Confusion Matrix:")
    print(f"Kalacakları Doğru Bildi (True Negative)   : {cm[0][0]}")
    print(f"Yanlış Alarm Verdi (False Positive)       : {cm[0][1]} (Kalacak kişiye ayrılır dedi)")
    print(f"KAÇIRILANLAR (False Negative - TEHLİKE!)  : {cm[1][0]} (Ayrılacak kişiye kalır dedi)")
    print(f"YAKALANANLAR (True Positive - BAŞARI!)    : {cm[1][1]} (Ayrılacak kişiyi doğru buldu)\n")
    
    # metrikler (Precision, Recall, F1)
    # classification_report, her iki sınıf (0 ve 1) için tüm metrikleri otomatik basar.
    print(classification_report(y_test, y_pred))
```

```python
# modelleri tanımlıyoruz (SVM için probability=True ekledik)
optimize_modeller = {
    "Gradient Boosting": GradientBoostingClassifier(learning_rate = 0.1, max_depth = 3, n_estimators = 200, random_state = 42),
    "Logistic Regression": LogisticRegression(C = 10, class_weight = "balanced", max_iter = 1000, random_state = 42),
    "Random Forest": RandomForestClassifier(class_weight = "balanced", max_depth = 15, n_estimators = 100, random_state = 42),
    "SVM (Destek Vektör)": SVC(C = 1, class_weight = "balanced", kernel = "rbf", random_state = 42, probability = True) # probability=True ekledik
}

yeni_esik_degeri = 0.35

for isim, model in optimize_modeller.items():

    model.fit(X_train_scaled, y_train)
    
    # modelin ürettiği olasılıkları çekme
    olasiliklar = model.predict_proba(X_test_scaled)[:, 1]
    
    # 0.35'ten büyük olanları 1 (ayrılır), küçükleri 0 (kalır) yap
    y_pred_yeni_esik = (olasiliklar >= yeni_esik_degeri).astype(int)
    
    # yeni confusion matrix hesapla
    cm_yeni = confusion_matrix(y_test, y_pred_yeni_esik)
    
    # sınıf 1 (ayrılanlar) için recall hesaplama : TP / (TP + FN)
    tp = cm_yeni[1][1]
    fn = cm_yeni[1][0]
    recall_skoru = tp / (tp + fn)
    
    print(f"{isim.upper()}")
    print(f"   Yakalananlar (Başarı)     : {tp} (Recall: %{recall_skoru*100:.0f})")
    print(f"   Kaçırılanlar (Tehlike)    : {fn}")
    print(f"   Yanlış Alarm (Maliyet)    : {cm_yeni[0][1]}")
    print("-" * 40)
```

```python
from sklearn.model_selection import cross_val_predict

# test edeceğimiz eşik değerleri
esikler = [0.25, 0.30, 0.35, 0.40, 0.50]
cv_sonuclari = []

# her model için cross validation
for isim, model in optimize_modeller.items():
    
    # cross_val_predict, modeli 5 parçada eğitip, her seferinde görmediği parçayı tahmin eder.
    # böylece eğitim setinde bile olasılıklar elde ederiz.
    y_cv_olasilik = cross_val_predict(model, X_train_scaled, y_train, cv=5, method='predict_proba', n_jobs=-1)[:, 1]
    
    # eşik değerlerini test etme
    for esik in esikler:
        # olasılığı eşiğe göre 0 veya 1'e çevirme
        y_cv_pred = (y_cv_olasilik >= esik).astype(int)
        
        # sınıf 1 (işten ayrılanlar) için metrikleri hesaplama
        precision = precision_score(y_train, y_cv_pred, zero_division=0)
        recall = recall_score(y_train, y_cv_pred, zero_division=0)
        f1 = f1_score(y_train, y_cv_pred, zero_division=0)
        
        cv_sonuclari.append({
            "Model": isim,
            "Eşik": esik,
            "Precision": round(precision, 3), # Yanlış alarmlar artarsa bu düşer
            "Recall": round(recall, 3),       # Yakalama oranımız
            "F1-Score": round(f1, 3)          # İkisinin dengesi
        })

cv_df = pd.DataFrame(cv_sonuclari)

for isim in optimize_modeller.keys():
    print(f"{isim.upper()} (CV Sonuçları)")
    print(f"{'='*40}")
    
    # Sadece o modele ait veriyi ekrana bas
    print(cv_df[cv_df["Model"] == isim].drop("Model", axis=1).to_string(index=False))
    print("\n")
```

```python
from sklearn.pipeline import Pipeline

esikler = [0.25, 0.30, 0.35, 0.40, 0.50]
cv_sonuclari = []

for isim, model in optimize_modeller.items():
    
    # pipeline kullanımını buraya ekledik
    # modele ham X_train vereceğiz, Pipeline her cross validation döngüsünde (5 kere) 
    # veriyi kendi içinde ölçeklendirip modele öyle sunacak.veri sızıntısının olabildiğince önüne geçmek amacımız.
    pipeline = Pipeline([
        ('scaler', StandardScaler()), 
        ('model', model)
    ])
    
    # X_train_scaled yerine, ham X_train kullandık
    y_cv_olasilik = cross_val_predict(
        pipeline, 
        X_train, # ham veriyi verdik
        y_train, 
        cv=5, 
        method='predict_proba', 
        n_jobs=-1
    )[:, 1]
    
    # eşik değerlerini test ediyoruz
    for esik in esikler:
        y_cv_pred = (y_cv_olasilik >= esik).astype(int)
        cv_sonuclari.append({
            "Model": isim,
            "Eşik": esik,
            "Precision": round(precision_score(y_train, y_cv_pred, zero_division=0), 3),
            "Recall": round(recall_score(y_train, y_cv_pred, zero_division=0), 3),
            "F1-Score": round(f1_score(y_train, y_cv_pred, zero_division=0), 3)
        })

cv_df = pd.DataFrame(cv_sonuclari)
for isim in optimize_modeller.keys():
    print(f"{isim.upper()}\n{'='*40}")
    print(cv_df[cv_df["Model"] == isim].drop("Model", axis=1).to_string(index=False))
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# denemelere ve kod çıktılarına göre logistic regression modelini kullanmayı tercih ettik ,pipeline oluşturduk.
# yeni bir veri geldiğinde ölçeklendirdik mi ? diye düşünmemek için.
final_model = Pipeline([
    ('scaler', StandardScaler()),
    ('logistic', LogisticRegression(C=10, class_weight='balanced', max_iter=1000, random_state=42))
])

# modeli eğitim verisi ile (X_train) eğitiyoruz
final_model.fit(X_train, y_train)

# hiç görmediği test veri setinde (X_test) olasılıkları tahmin ediyoruz.
# burda artık X_test_scaled değil , ham X_test veriyoruz. pipeline kısmında işi kendi çözecek.
test_olasiliklar = final_model.predict_proba(X_test)[:, 1]

# belirlediğimiz eşik(threshold) değerini uyguluyoruz
belirlediğimiz_eşik = 0.35
y_test_pred = (test_olasiliklar >= belirlediğimiz_eşik).astype(int)

# sonuçları yazdırma
print(f"X_test üzerindeki performansı")
print(f"{'-'*50}")

print(f"\nSeçtiğimiz Model: Logistic Regression")
print(f"Kullanılan Eşik Değeri: {belirlediğimiz_eşik}\n")

cm_final = confusion_matrix(y_test, y_test_pred)
print("Confusion Matrix:")
print(f"Kalacakları Doğru Bildi (True Negative)   : {cm_final[0][0]}")
print(f"Yanlış Alarm Verdi (False Positive)       : {cm_final[0][1]} (Maliyet)")
print(f"KAÇIRILANLAR (False Negative - TEHLİKE!)  : {cm_final[1][0]}")
print(f"YAKALANANLAR (True Positive - BAŞARI!)    : {cm_final[1][1]}\n")

print("sınıflandırma performansı")
print(classification_report(y_test, y_test_pred))

# görsel açıdan görmek için confusion matrixi (Isı Haritası)
plt.figure(figsize=(8, 6))
sns.heatmap(cm_final, annot=True, fmt='d', cmap='Blues', 
            xticklabels=["0", "1"], 
            yticklabels=["0", "1"])
plt.title(f"Confusion Matrisimiz (Eşik: {belirlediğimiz_eşik})", fontsize=14, pad=15)
plt.xlabel("Modelin Tahmini", fontsize=12)
plt.ylabel("Gerçek Durum", fontsize=12)
plt.tight_layout()
plt.show()
```
