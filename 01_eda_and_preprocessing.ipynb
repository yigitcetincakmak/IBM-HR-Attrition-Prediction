```python
Tr = { 
    
    "is_seyahati_sikligi" : { "Travel_Rarely" : "Nadir Seyahat",
                             "Travel_Frequently" : "Sık Seyahat", 
                             "Non-Travel" : "Seyahat Yok" },


    "egitim_alani" : {
                    "Life Sciences": "Yaşam Bilimleri",
                    "Other": "Diğer",
                    "Medical": "Sağlık",
                    "Marketing": "Pazarlama",
                    "Technical Degree": "Teknik Bölüm",
                    "Human Resources": "İnsan Kaynakları"},

    "cinsiyet" : {
                 "Female": "Kadın",
                 "Male": "Erkek"  },


    "is_rolu" : {
          "Sales Executive": "Satış Yöneticisi",
          "Research Scientist": "Araştırma Uzmanı",
          "Laboratory Technician": "Laboratuvar Teknisyeni",
          "Manufacturing Director": "Üretim Direktörü",
          "Healthcare Representative": "Sağlık Temsilcisi",
          "Manager": "Müdür",
          "Sales Representative": "Satış Temsilcisi",
          "Research Director": "Araştırma Direktörü",
          "Human Resources": "İnsan Kaynakları" },

   
    "medeni_hal" : {
         "Single": "Bekar",
         "Married": "Evli",
         "Divorced": "Boşanmış" }

}


for sutun_adı , degerler in Tr.items() :
    if sutun_adı in df.columns:
        df[sutun_adı] = df[sutun_adı].map(degerler)



    
```

```python
seyahat_sıklıkları = df["is_seyahati_sikligi"].value_counts() # id ye gerek yok tek tek saymak için
seyahat_sıklıkları
```

```python
departman_df = df["departman"].value_counts()
departman_df = departman_df.reset_index()
departman_df_oran = df["departman"].value_counts(normalize=True).reset_index().round(2)

departman_df_merge = pd.merge(departman_df,departman_df_oran , on="departman")
departman_df_merge = departman_df_merge.rename({"count":"Kişi Sayısı" , "proportion" : "% Olasılık"} , axis=1)

departman_df_merge
```

```python
grafik_departman = plt.barh(  
                    
                    y = departman_df_merge["departman"],
                    width = departman_df_merge["Kişi Sayısı"], 
                    color = ["red","green","blue"]
                
                 )


plt.ylabel("Departman", fontsize=14, fontweight="bold" , color = "gray")
plt.xlabel("Kişi Sayısı" , fontsize=14, fontweight="bold" , color = "gray")

plt.yticks(fontsize = 10 , fontweight = "bold")
plt.xticks(fontsize = 10 , fontweight = "bold")



ozel_etiketler = []
kisi_sutunu = departman_df_merge["Kişi Sayısı"]
olasilik_sutunu = departman_df_merge["% Olasılık"]

for kisi, olasilik in zip(kisi_sutunu, olasilik_sutunu):
    yuzde_degeri = olasilik * 100
    hazirlanan_metin = f"{kisi}\n(%{yuzde_degeri:.0f})"
    ozel_etiketler.append(hazirlanan_metin)

# fmt='%g' yerine labels=ozel_etiketler parametresini kullanıyoruz olasılığıda çubukta kişi sayısının altına yazdırmak için
plt.bar_label(grafik_departman, labels=ozel_etiketler, padding=5, fontweight="bold", fontsize=11, color = "black")



#for container in grafik.containers:
 #   grafik.bar_label(container, fmt='%.2f', padding=5, fontweight="bold")

# plt.bar_label(grafik, fmt='%g', padding=5, fontweight="bold") # matplotlip'te döngüye gerek yok


plt.grid(axis="x")

plt.xlim(0,1150)
plt.tight_layout()
plt.show()
```

```python
df_egitim_alan = df["egitim_alani"].value_counts()
df_egitim_alan = df_egitim_alan.reset_index()

df_egitim_alan = df_egitim_alan.rename({"egitim_alani" : "Eğitim_Alanı","count" : "Kişi_Sayısı"} , axis=1)

df_egitim_alan
```

```python
# grafik

renkler = sns.color_palette("pastel")

# 6 dilim için 6 rakam. (0.1) gib değerler ile dışarı itiyoruz.
ayirma_mesafeleri = [0.1, 0.1, 0.1, 0.1, 0.2, 0.3,]

# seabornun pasta grafiği yok.
plt.pie(
   
    x = df_egitim_alan["Kişi_Sayısı"], 
    labels = df_egitim_alan["Eğitim_Alanı"],
    colors = renkler,
    autopct='%.1f%%',# yüzdelik dilimleri otomatik yazıyor.
    wedgeprops={'edgecolor': 'black'}, # dilimler arası çizgi.
    explode=ayirma_mesafeleri,
   
    )
plt.title("Eğitim Alanına Göre Kişi Dağılımı" , 
          fontweight = "bold" , 
          fontsize = 12 , 
          x=0.6) # x=0.6,başlığı x ekseninde hareket ettir.
```
